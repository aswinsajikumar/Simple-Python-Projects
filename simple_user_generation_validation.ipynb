## Project: User Registration System with Validation and Password Strength Checker



**Project Overview:** In this project, you will develop a Python program that simulates a user registration system. The program will allow users to register by providing their username and password. It will include validation checks for username uniqueness and password strength assessment. The project aims to enhance understanding of string manipulation, regular expressions, and basic data handling in Python.



**Project Requirements:**

* User Registration: Prompt the user to enter a username and password for registration.

* Username Uniqueness: Ensure the username is unique by checking against previously registered users.

* Password Strength Checker: Implement a function that evaluates the strength of the password based on specific criteria (e.g., length, inclusion of special characters, numbers).

* Data Storage: Store registered users in a list or dictionary for future reference.

* Feedback to User: Provide appropriate messages based on the registration outcome (success or failure).

**Example Flow:**

* The user is prompted to enter a username and password.

* The program checks if the username is already taken.

* The program evaluates the strength of the password.

* If both checks pass, the user is registered successfully; otherwise, appropriate error messages are displayed.



```python
# Import regular expression module
import re
```

```python
# List to store user data
reg_users = []
```

```python
# Function to check password strength and show missing conditions
def str_password(password):
    errors = []

    if len(password) < 8:
        errors.append("Password must be at least 8 characters long.")
    if not re.search(r"[a-z]", password):
        errors.append("Password must include at least one lowercase letter.")
    if not re.search(r"[A-Z]", password):
        errors.append("Password must include at least one uppercase letter.")
    if not re.search(r"[0-9]", password):
        errors.append("Password must include at least one number.")
    if not re.search(r"[!@#$%^&*(),.?\":{}|<>]", password):
        errors.append("Password must include at least one special character (!@#$%^&*(),.?\":{}|<>).")

    if errors:
        for e in errors:
            print(e)
        return False
    return True
```

```python
# New user registration
def new_user(username, password):
    if not str_password(password):
        print("Password is too weak")
        return False
    reg_users.append(username)
    print(f"User '{username}' registered successfully!")
    return True
```

```python
# Main loop
while True:
    username = input("Enter your username: ")

    # Check if username is already taken
    if username in reg_users:
        print("Username is already taken. Please choose a different one.")
        continue

    password = input("Enter your password: ")
    if new_user(username, password):
        break
```
